```python
print("Olá vamos a aplicação?")

print("MENU")
print("(1) Criar turma")
print("(2) Inserir um aluno na turma")
print("(3) Listar a turma")
print("(4) Consultar um aluno por id")
print("(5) Guardar a turma num ficheiro")
print("(6) Carregar uma turma dum ficheiro")
print("(0) Sair da aplicação")

a = int(input("Introduza um numero de 0 a 5"))

while a!=0:
    print("MENU")
    print("(1) Criar turma")
    print("(2) Inserir um aluno na turma")
    print("(3) Listar a turma")
    print("(4) Consultar um aluno por id")
    print("(5) Guardar a turma num ficheiro")
    print("(6)Carregar uma turma dum ficheiro")
    print("(0) Sair da aplicação")

    a = int(input("Introduza um numero de 0 a 5"))
    if a == 1:
        turma=[]

    elif a==2:
        notas = []
        print("Insira os dados do aluno")
        nome = input("Insira o nome")
        id = input("Insira o id")
        notaTPC = float(input("Insira a nota do TPC"))
        notaProj = float(input("Insira a nota do projeto"))
        notaTeste = float(input("Insira a nota do teste"))
        notas.append(notaTPC)
        notas.append(notaProj)
        notas.append(notaTeste)
        aluno=(nome,id,notas)
        turma.append(aluno)

    elif a == 3:
        for aluno in turma:
            print(aluno)
    
    elif a == 4:
        print("Qual o id desejado?")
        x = input("Insira o id")
        for aluno in turma:
            if aluno[1] == x:
                print(aluno)

    elif a == 5:
        f = open("turma.csv","w")
        for aluno in turma:
         linha = (f"{aluno[0]},{aluno[1]},{aluno[2][0]},{aluno[2][1]},{aluno[2][2]}")
         f.write(linha)
         f.write("\n")
        f.close()

    elif a ==6:
        turma=[]
        f = open("turma.csv")
        alunos = f.readlines()
        for aluno in alunos:
            dado = aluno.split(",")
            turma.append((dado[0],dado[1],[float(dado[2]),float(dado[3]),float(dado[4])]))
        f.close()
        print(turma)

print("A aplicação está encerrada, até mais!")

```
